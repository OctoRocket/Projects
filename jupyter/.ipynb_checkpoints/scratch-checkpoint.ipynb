First, define a recursive square root function.

```haskell
intSqrt :: Integer -> Integer
intSqrt = worker 1
    where 
        worker :: Integer -> Integer -> Integer
        worker n m = if ((n + 1) * (n + 1)) > m
            then n
            else worker (n + 1) m
```

```haskell
intSqrt 16
```

Next, define the main factor function.

```haskell
factors :: Integer -> [(Integer, Integer)]
factors n = map (\x -> (x, div n x))
    $ filter ((==) 0 . mod n)
    [1..intSqrt n]
```

```haskell
factors 68
```
